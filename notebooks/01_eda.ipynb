# Notebook 01 — Exploratory Data Analysis

This notebook inspects both raw datasets before any cleaning or transformation.

- **Dataset 1:** `AI-based Career Recommendation System.csv` (~200 rows)
- **Dataset 2:** `career_dataset_large.xlsx` (~5 000 rows)

---
## Dataset 1 — `AI-based Career Recommendation System.csv`

```python
import pandas as pd

ds1_path = '../data/raw/AI-based Career Recommendation System.csv'
df1 = pd.read_csv(ds1_path)

print('=== Dataset 1 — Basic Info ===')
print(f'Shape: {df1.shape}')
print()
print('Column names:')
print(df1.columns.tolist())
print()
print('Data types:')
print(df1.dtypes)
print()
print('Null counts per column:')
print(df1.isnull().sum())
print()
print(f'Duplicate row count: {df1.duplicated().sum()}')
```

```python
print('=== Dataset 1 — Career Label Distribution ===')
career_col = 'Recommended_Career'
print(df1[career_col].value_counts())
print()
print(f'Unique career count: {df1[career_col].nunique()}')
```

```python
print('=== Dataset 1 — Top-20 Skills by Frequency ===')
top_skills = (
    df1['Skills']
    .dropna()
    .str.split(';')
    .explode()
    .str.strip()
    .value_counts()
    .head(20)
)
print(top_skills)
```

```python
print('=== Dataset 1 — Top-20 Interests by Frequency ===')
top_interests = (
    df1['Interests']
    .dropna()
    .str.split(';')
    .explode()
    .str.strip()
    .value_counts()
    .head(20)
)
print(top_interests)
```

```python
print('=== Dataset 1 — Education Level Distribution ===')
print(df1['Education'].value_counts())
```

## Dataset 2 — `career_dataset_large.xlsx`

```python
ds2_path = '../data/raw/career_dataset_large.xlsx'
df2 = pd.read_excel(ds2_path, engine='openpyxl')

print('=== Dataset 2 — Basic Info ===')
print(f'Shape: {df2.shape}')
print()
print('Column names:')
print(df2.columns.tolist())
print()
print('Data types:')
print(df2.dtypes)
print()
print('Null counts per column:')
print(df2.isnull().sum())
print()
print(f'Duplicate row count: {df2.duplicated().sum()}')
```

```python
print('=== Dataset 2 — Career Label Distribution ===')
career_col2 = 'Recommended Career'
print(df2[career_col2].value_counts())
print()
print(f'Unique career count: {df2[career_col2].nunique()}')
```

```python
print('=== Dataset 2 — Top-20 Skills by Frequency ===')

# Detect delimiter: check first non-null value for ';' vs ','
first_skill_val = df2['Skills'].dropna().iloc[0]
skill_delimiter = ';' if ';' in first_skill_val else ','
print(f'Detected skill delimiter: {repr(skill_delimiter)}')
print()

top_skills2 = (
    df2['Skills']
    .dropna()
    .str.split(skill_delimiter)
    .explode()
    .str.strip()
    .value_counts()
    .head(20)
)
print(top_skills2)
```

```python
print('=== Dataset 2 — Education Level Distribution ===')
print(df2['Education Level'].value_counts())
```

```python
print('=== Dataset 2 — Specialization Cardinality and Top-20 Values ===')
spec_col = 'Specialization'
print(f'Total unique specializations: {df2[spec_col].nunique()}')
print()
print('Top-20 specializations by frequency:')
print(df2[spec_col].value_counts().head(20))
```
